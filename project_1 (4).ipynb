```python
class EmployeDetails:
    # contructor function
    def __init__(self, name="user", basic_salary=0, hra=0, da=0,pf = 0,tax=0):
        # class Attributes(variables,data)
        self.name = name
        self.basic_salary = basic_salary
        self.hra = hra
        self.da = da
        self.pf = pf
        self.tax = tax
        self.total_salary = self.basic_salary + self.hra + self.da - self.pf - self.tax

    # class methods(functions)
    def calculate_salary(self):
        pass

    def generate_slip(self):
        print("------ SALARY SLIP ------")
        print("Employee Name:", self.name)
        print("Basic Salary:", self.basic_salary)
        print("HRA:", self.hra)
        print("DA:", self.da)
        print("PF:", self.pf)
        print("Tax:", self.tax)
        print("Net Salary:", self.total_salary)
```

```python
#creating an object
E1 =EmployeDetails("Masoom",basic_salary=50000,hra = 5000,da = 2000,pf = 1000,tax = 500)
print(E1.name)
print(E1.basic_salary)
print(E1.hra)
print(E1.da)
print(E1.pf)
print(E1.tax)
print(E1.total_salary)
```
